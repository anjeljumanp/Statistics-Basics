1. What is the difference between descriptive statistics and inferential statistics? Explain with examples.
  
  * **Descriptive Statistics**

    - Meaning: Descriptive statistics deals with collecting, organizing, summarizing, and presenting data in a meaningful way. It does not go beyond the given data.
    - Purpose: To describe the main features of a dataset clearly and simply.

    - Main tools used:
      - Measures of central tendency: Mean, Median, Mode
      - Measures of dispersion: Range, Variance, Standard Deviation
      - Tables, charts, graphs, percentages

    - Example:
      - Calculating the average marks of 50 students in an economics exam.
      - Presenting the monthly rainfall data of Kerala using a bar diagram.
      - Finding the average income of employees in a company based on collected data.

  * **Inferential Statistics**

    - Meaning: Inferential statistics involves drawing conclusions, predictions, or generalizations about a population based on sample data.
    - Purpose: To make inferences or decisions about a large group when complete data is not available.
    - Main tools used:
      - Sampling techniques
      - Estimation (confidence intervals)
      - Hypothesis testing (t-test, z-test, chi-square test, ANOVA)
    - Example:
      - Surveying 200 voters and predicting the election result of an entire state.
      - Testing whether a new teaching method improves student performance using sample results.
      - Estimating the average income of all households in India based on a sample survey.

2. What is sampling in statistics? Explain the differences between random and stratified sampling.

 -**Sampling in Statistics**
   - Meaning: Sampling is the process of selecting a representative subset (sample) from a large group (population) in order to study its characteristics and draw conclusions about the entire population.
   - Purpose of sampling:
     - Saves time, cost, and effort
     - Makes study possible when the population is very large
     - Provides reliable results when done scientifically
    - Example: Instead of surveying all voters in India, a researcher studies a sample of 6,000 voters to predict election trends.

- Random Sampling
  - In random sampling, each unit of the population has an equal and independent chance of being selected. The selection is made purely by chance, such as through the lottery method or random numbers. This method is simple and free from personal bias, but it may not adequately represent all subgroups of the population, especially when the population is heterogeneous.
- Stratified Sampling
  - In stratified sampling, the population is first divided into homogeneous subgroups called strata based on characteristics like age, gender, income, or education. Samples are then selected randomly from each stratum, often in proportion to their size. This method ensures proper representation of all important subgroups and provides more accurate results.

3. Define mean, median, and mode. Explain why these measures of central tendency are important.
 - Mean:
   - The mean is the arithmetic average of a set of values. It is obtained by adding all observations and dividing by the number of observations.
   - Example: Marks: 40, 50, 60.   Mean = (40 + 50 + 60) / 3 = 50

 - Median:
    - The median is the middle value of a dataset when the observations are arranged in ascending or descending order.
    - If the number of observations is odd, the median is the middle value. - If it is even, the median is the average of the two middle values.
    - Example: Data: 10, 20, 30, 40, 50.  Median = 30

 - Mode:
    - The mode is the value that occurs most frequently in a dataset.
    - Example: Data: 2, 4, 4, 6, 8.  Mode = 4

 - Importance of Measures of Central Tendency
   - Summarize large data: They reduce a large mass of data into a single representative value, making interpretation easy.
   - Facilitate comparison: They help compare two or more datasets, such as average income of different regions.
   - Aid decision-making: Used in economics, business, and government for planning and policy formulation.
   - Basis for further analysis: They form the foundation for advanced statistical tools like dispersion, correlation, and regression.
   - Different situations require different measures:
     - Mean is suitable when data is uniform.
     - Median is useful when data contains extreme values.
     - Mode is useful for identifying the most common value.

4. Explain skewness and kurtosis. What does a positive skew imply about the data?

  - **Skewness** refers to the degree of asymmetry in a frequency distribution. A distribution is said to be symmetrical when the values are evenly distributed on both sides of the central value, and in such a case the mean, median, and mode coincide. When the distribution is not symmetrical, it is skewed.
  - Skewness can be either positive or negative. In a positively skewed (right-skewed) distribution, the tail of the distribution extends towards the right side, indicating the presence of a few very high values. In such a distribution, most observations are concentrated at the lower end, and the mean is greater than the median, which in turn is greater than the mode. In a negatively skewed (left-skewed) distribution, the tail extends towards the left side due to a few very low values, and the mean is less than the median and mode.
  - **Kurtosis** refers to the degree of peakedness or flatness of a distribution in comparison with a normal distribution. It indicates how closely the values are clustered around the mean and the nature of the tails. A leptokurtic distribution is highly peaked with a large number of observations near the mean and relatively heavy tails. A mesokurtic distribution represents a normal distribution with a moderate peak and average tails. A platykurtic distribution is flatter, showing that the data are more widely spread with lighter tails.
  - **A positive skew** implies that the distribution has a longer right tail caused by a small number of high extreme values. This suggests that most of the data values are relatively low, while a few high values increase the average. Income distribution is a common example of positive skewness, where a majority of people earn lower incomes and a few earn exceptionally high incomes.

5.  Implement a Python program to compute the mean, median, and mode of a given list of numbers.
  numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

```python
from statistics import mean, median, mode

# Given list of numbers
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

# Calculations
mean_value = mean(numbers)
median_value = median(numbers)
mode_value = mode(numbers)

# Output
print("Numbers:", numbers)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

6. Compute the covariance and correlation coefficient between the following two datasets provided as lists in Python:
 - list_x = [10, 20, 30, 40, 50]
 - list_y = [15, 25, 35, 45, 60]

```python
import numpy as np

# Given lists
list_x = np.array([10, 20, 30, 40, 50])
list_y = np.array([15, 25, 35, 45, 60])

# Covariance (sample covariance)
covariance = np.cov(list_x, list_y)[0][1]

# Correlation coefficient
correlation = np.corrcoef(list_x, list_y)[0][1]

# Output
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```

7. Write a Python script to draw a boxplot for the following numeric list and identify its outliers. Explain the result:
- data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

```python
import matplotlib.pyplot as plt
import numpy as np

# Given data
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

# Create boxplot
plt.boxplot(data)
plt.title("Boxplot of Given Data")
plt.ylabel("Values")
plt.show()

# Convert to numpy array
data_np = np.array(data)

# Quartiles
Q1 = np.percentile(data_np, 25)
Q3 = np.percentile(data_np, 75)
IQR = Q3 - Q1

# Outlier limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Identify outliers
outliers = data_np[(data_np < lower_limit) | (data_np > upper_limit)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:", outliers)
```

8.  You are working as a data analyst in an e-commerce company. The marketing team wants to know if there is a relationship between advertising spend and daily sales.
- Explain how you would use covariance and correlation to explore this relationship.
- Write Python code to compute the correlation between the two lists:
- advertising_spend = [200, 250, 300, 400, 500]   
- daily_sales = [2200, 2450, 2750, 3200, 4000]

As a data analyst, I would first use covariance to understand the direction of the relationship between advertising spend and daily sales.

If covariance is positive, it means higher advertising spend is associated with higher sales.

If covariance is negative, higher ad spend is associated with lower sales.

If covariance is zero, there is no linear relationship.

However, covariance alone is hard to interpret because its value depends on the units of measurement (rupees, dollars, number of sales, etc.).

To overcome this limitation, I would use correlation.

Correlation coefficient (r) measures both the strength and direction of the relationship.
 - Its value lies between −1 and +1.
 - +1 → perfect positive relationship
 - 0 → no linear relationship
 - −1 → perfect negative relationship

Correlation helps the marketing team clearly understand how strongly advertising spend influences daily sales.

```python
import numpy as np

# Given data
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

# Convert to numpy arrays
ad_spend = np.array(advertising_spend)
sales = np.array(daily_sales)

# Compute correlation coefficient
correlation = np.corrcoef(ad_spend, sales)[0, 1]

print("Correlation coefficient:", correlation)
```

9. Your team has collected customer satisfaction survey data on a scale of 1-10 and wants to understand its distribution before launching a new product.
- Explain which summary statistics and visualizations (e.g. mean, standard deviation, histogram) you’d use.
- Write Python code to create a histogram using Matplotlib for the survey
- data: survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

Before launching a new product, it’s important to understand how customer satisfaction scores are distributed. I would use the following:

1. Summary Statistics

Mean: Shows the average satisfaction level. Useful to know overall customer sentiment.

Median: Indicates the central score and is helpful if the data is slightly skewed.

Standard Deviation: Measures variability in responses. A low value means customers feel similarly; a high value indicates mixed opinions.

Minimum and Maximum: Show the range of satisfaction (worst and best experiences).

Together, these statistics summarize the dataset into meaningful numbers for quick decision-making.

2. Visualizations

Histogram:

Shows the shape of the distribution (symmetric, skewed, clustered).

Helps identify whether most customers are satisfied (scores closer to 8–10) or dissatisfied (scores closer to 1–4).

Useful for spotting gaps or unusual patterns in responses.

```python
import matplotlib.pyplot as plt

# Survey data
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

# Create histogram
plt.hist(survey_scores, bins=7)
plt.title("Histogram of Customer Satisfaction Scores")
plt.xlabel("Satisfaction Score")
plt.ylabel("Frequency")
plt.show()
```
